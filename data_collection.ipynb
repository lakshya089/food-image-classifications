```python
from google_play_scraper import Sort, reviews,reviews_all
import pandas as pd
```

```python
# initialization

app_name = 'org.telegram.messenger' # found in app's url

rvws, token = reviews(
    app_name,           # found in app's url
    lang='en',        # defaults to 'en'
    country='us',     # defaults to 'us'
    sort=Sort.NEWEST, # start with most recent
    count=200       # batch size, set as 200 since 200 reviews per page
)

app_reviews = rvws
batch_num=5
```

```python
# Loop through the number of batches
for batch in range(1000): # You can adjust the number of batches to collect more or less reviews
    rvws, token = reviews( # store continuation_token
        app_name,
        lang='en',
        country='us',
        sort=Sort.NEWEST,
        count=200,
        # using token obtained from previous batch
        continuation_token=token
    )
    app_reviews.extend(rvws)
    batch_num +=1
```

```python
df = pd.DataFrame.from_records(app_reviews)
df_data = df[['content','score']]
```

```python
df_data.to_csv('telegram.csv',index=False)
```
